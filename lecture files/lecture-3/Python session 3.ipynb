## 2. NumPy arrays

In previous lessons, we used Python lists to store sensor readings.

NumPy provides a specialised data structure called an **array**. Arrays are particularly useful for numerical calculations because operations can be applied to many values at once.

Let's compare a Python list with a NumPy array.

````python
# A normal Python list
angle_list = [21.5, 72.1, 32.8, 63.0, 22.4]

# A NumPy array
angle_array = np.array(angle_list)

print("Python list:", angle_list)
print("NumPy array:", angle_array)

print("Array type:", type(angle_array))
````

### Indexing and slicing

Like Python lists, NumPy arrays use zero-based indexing.

- `array[0]` returns the first element.
- `array[-1]` returns the last element.
- `array[1:4]` selects elements at indices 1, 2, and 3.

The end index of a slice is not included.

````python
print("First reading:", angle_array[0])
print("Last reading:", angle_array[-1])

print("First three readings:", angle_array[:3])
print("Readings at indices 1 through 3:", angle_array[1:4])
````

### Array dimensions and shapes

A NumPy array has a shape that describes its dimensions.

For example:

- A one-dimensional array stores a sequence of values.
- A two-dimensional array can represent rows and columns, such as readings from multiple sensors.

Important properties include:

- `shape`: the length along each dimension.
- `ndim`: the number of dimensions.
- `size`: the total number of elements.

````python
print("Shape:", angle_array.shape)
print("Dimensions:", angle_array.ndim)
print("Number of elements:", angle_array.size)

# A two-dimensional array: three samples from two sensors
sensor_data = np.array([
    [21.5, 450],
    [22.1, 470],
    [22.8, 490]
])

print("\nSensor data:")
print(sensor_data)

print("Shape:", sensor_data.shape)
print("Dimensions:", sensor_data.ndim)
print("Number of elements:", sensor_data.size)

# Select the second sample
print("Second sample:", sensor_data[1])

# Select all samples from the first sensor
print("First sensor readings:", sensor_data[:, 0])
````

## 3. Vectorized operations and conversions

One major advantage of NumPy is that we can perform calculations on an entire array without **writing a loop for every element**.

This is called a **vectorized operation**.

For example, suppose we have multiple ADC readings from a sensor. We want to convert these values into voltage.
The ADC scaling formula here is illustrative. Actual ESP32 ADC readings depend on the device configuration, attenuation, and calibration.


angle ipv voltage

````python
adc_readings = np.array([500, 1000, 2000, 3000, 4000])

reference_voltage = 3.3
max_adc = 4095

voltages = adc_readings / max_adc * reference_voltage

print("ADC readings:", adc_readings)
print("Voltages:", voltages)
````

### Compare a loop with a vectorized operation

In ordinary Python (or C), you might convert each reading using a loop. With NumPy, you can apply the same formula to the entire array.

NumPy also supports:

- Addition and subtraction
- Multiplication and division
- Comparisons
- Boolean masks for selecting data

## 4. NumPy statistics and filtering

Sensor readings often contain noise and fluctuations. NumPy provides convenient functions for summarizing numerical data.

Common statistical functions:

| Function | Purpose |
|---|---|
| `np.mean()` | Calculate the average |
| `np.min()` | Find the minimum |
| `np.max()` | Find the maximum |
| `np.std()` | Calculate standard deviation |
| `np.median()` | Find the median |

Standard deviation describes how spread out values are around their mean. NumPy's default `np.std()` uses the population standard deviation (`ddof=0`) e.g. your readings represent the entire dataset you want to describe.

````python
angles = np.array([
    71.5, 54.1, 42.8, 23.0, 72.4,
    52.7, 83.2, 81.9, 12.5, 124.0
])

print("Mean:", np.mean(angles))
print("Minimum:", np.min(angles))
print("Maximum:", np.max(angles))
print("Standard deviation:", np.std(angles))
print("Median:", np.median(angles))
````

### Filtering values with conditions

We can use a condition to create a Boolean mask. The mask can then select only the values that satisfy that condition.

For example, we may want to identify temperatures above a threshold or exclude readings that are outside an acceptable range.

````python
threshold = 23.0

# A Boolean mask
mask = angles > threshold

print("Mask:", mask)

# Select only readings above the threshold
high_readings = angles[mask]

print("Readings above threshold:", high_readings)
print("Number of high readings:", len(high_readings))
````

## 5. Matplotlib — visualizing sensor data

Matplotlib allows us to turn numerical data into graphs.

A line plot is useful for showing how a measurement changes over time. A histogram is useful for examining the distribution of measurements.

A readable graph should normally include:

- A title
- An x-axis label
- A y-axis label
- A legend when multiple series are shown
- A grid when it improves readability

````python
sample_number = np.arange(len(angles))

plt.plot(
    sample_number,
    angles,
    marker="o",
    label="Angle"
)

plt.xlabel("Sample number")
plt.ylabel("Angle (°)")
plt.title("Angle Sensor Readings")
plt.legend()
plt.grid(True)
plt.show()
````

### Histograms

A histogram groups numerical values into intervals called bins. It helps us see where measurements are concentrated and whether there are unusual values.

Unlike a line plot, a histogram does not show the order in which readings were taken.

````python
plt.hist(angles, bins=5, edgecolor="black")

plt.xlabel("Angles (°)")
plt.ylabel("Frequency")
plt.title("Distribution of Angles Readings")
plt.grid(axis="y")
plt.show()
````

### Plot multiple series

We can plot both the original measurements and a reference threshold on the same graph.

This makes it easier to identify readings that exceed the allowed temperature.

````python
sample_number = np.arange(len(angles))
threshold = 45.0

plt.plot(
    sample_number,
    angles,
    marker="o",
    label="Measured angle"
)

plt.axhline(
    threshold,
    linestyle="--",
    label="Threshold"
)

plt.xlabel("Sample number")
plt.ylabel("Angle (°)")
plt.title("Angle and Threshold")
plt.legend()
plt.grid(True)
plt.show()
````

## 6. Integrated exercise — load and analyze sensor data

Now we will combine concepts from Classes 2 and 3.

Our workflow is:

1. Load readings from `readings.csv` if the file exists.
2. If the file is missing, generate simulated sensor data.
3. Convert the readings into a NumPy array.
4. Calculate standard statistics.
5. Identify readings above a threshold.
6. Plot the measurements and their distribution.

The CSV file is expected to contain a column named `angle`, as in the previous lesson.

This exercise uses the standard-library modules `csv`, `random`, and `pathlib`, along with NumPy and Matplotlib.

````python
import csv
import random
from pathlib import Path


def load_angle_data(filename):
    readings = []

    with open(filename, "r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            readings.append(float(row["angle"]))

    if not readings:
        raise ValueError("The CSV file contains no angle readings.")

    return np.array(readings, dtype=float)


csv_path = Path("readings.csv")

if csv_path.exists():
    try:
        angles = load_angle_data(csv_path)
        print(f"Loaded {len(angles)} readings from {csv_path}")

    except (OSError, ValueError, KeyError) as error:
        print("Could not load the CSV:", error)
        print("Generating simulated readings instead.")

        random.seed(7)
        angles = np.array([
            random.uniform(-180, 180) for _ in range(100)
        ])

else:
    print("readings.csv not found; generating simulated readings.")

    random.seed(7)
    angles = np.array([
        random.uniform(-180, 180) for _ in range(100)
    ])

print("First 10 readings:", angles[:10])
````

````python
# Calculate statistics for the loaded or simulated readings.

print("Number of readings:", angles.size)
print(f"Mean: {np.mean(angles):.2f} °")
print(f"Minimum: {np.min(angles):.2f} °")
print(f"Maximum: {np.max(angles):.2f} °")
print(f"Standard deviation: {np.std(angles):.2f} °")

threshold = 45.0

high_readings = angles[angles > threshold]

print("Readings above threshold:", len(high_readings))
print("Values above threshold:", high_readings)
````

````python
# Plot the sensor readings.

sample_number = np.arange(len(angles))

plt.figure(figsize=(10, 4))

plt.plot(
    sample_number,
    angles,
    label="Angle",
    linewidth=1.5
)

plt.axhline(
    30,
    linestyle="--",
    label="High angle threshold"
)

plt.xlabel("Sample number")
plt.ylabel("Angle (°)")
plt.title("Sensor Angle Over Time")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````

````python
# Plot the distribution of sensor readings.

plt.figure(figsize=(8, 4))

plt.hist(angles, bins=10, edgecolor="black")

plt.xlabel("Angle (°)")
plt.ylabel("Frequency")
plt.title("Distribution of Sensor Angle")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
````

### What is a merge conflict?

A merge conflict can occur when two sets of changes modify the same part of a file and Git cannot automatically decide which change to keep.

In VS Code:

1. Pull or merge the changes.
2. Open the file marked as conflicted.
3. Look for the conflict markers:
   - `<<<<<<< HEAD` marks the current version.
   - `=======` separates the versions.
   - `>>>>>>> ...` marks the incoming version.
4. Review both versions and choose the correct content, or combine them.
5. Remove all conflict markers.
6. Save the file.
7. Stage the resolved file, commit the resolution, and push.

VS Code may display options such as **Accept Current Change**, **Accept Incoming Change**, and **Accept Both Changes**.

For a Jupyter notebook, avoid resolving conflicts by blindly replacing the entire notebook JSON. If two people edit the same notebook, coordinating changes or assigning each person a different notebook often makes conflicts easier to prevent.
